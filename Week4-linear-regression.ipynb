```python
# Select columns that exist in this version of the dataset.
candidate_features = [
    "age", "studytime", "failures", "absences",
    "sex", "school", "address"
]
features = [c for c in candidate_features if c in df.columns]
print("Features used:", features)

X_reg = df[features].copy()
y_reg = df["G3"].copy()

categorical_features = X_reg.select_dtypes(include=["object", "category", "bool"]).columns.tolist()
numeric_features = [c for c in X_reg.columns if c not in categorical_features]

preprocessor_reg = ColumnTransformer(
    transformers=[
        ("categorical", OneHotEncoder(handle_unknown="ignore"), categorical_features),
        ("numeric", StandardScaler(), numeric_features)
    ],
    remainder="drop"
)

X_train, X_test, y_train, y_test = train_test_split(
    X_reg, y_reg, test_size=0.2, random_state=RANDOM_STATE
)

linear_model = make_pipeline(preprocessor_reg, LinearRegression())
linear_model.fit(X_train, y_train)
reg_predictions = linear_model.predict(X_test)

mae = mean_absolute_error(y_test, reg_predictions)
rmse = np.sqrt(mean_squared_error(y_test, reg_predictions))
r2 = r2_score(y_test, reg_predictions)

print(f"MAE:  {mae:.3f}")
print(f"RMSE: {rmse:.3f}")
print(f"R²:   {r2:.3f}")
```

```python
# Plot actual vs predicted grades
plt.figure(figsize=(6, 6))
plt.scatter(y_test, reg_predictions, alpha=0.7)
low = min(y_test.min(), reg_predictions.min())
high = max(y_test.max(), reg_predictions.max())
plt.plot([low, high], [low, high], linestyle="--")
plt.title("Linear Regression: Actual vs Predicted Grades")
plt.xlabel("Actual G3")
plt.ylabel("Predicted G3")
plt.tight_layout()
plt.savefig(os.path.join(GRAPH_DIR, "05_linear_regression_actual_vs_predicted.png"), dpi=150)
plt.show()

regression_results = pd.DataFrame({
    "Actual_G3": y_test.values,
    "Predicted_G3": reg_predictions
})
display(regression_results.head(10))
```
